```python
import pandas as pd
import numpy as np
```

```python
data ={
    'Name':['Amit','Bebo','Chetan','Amit','Divya','Ravi','Sneha','Kavya','Rahul','Pooja','Manish','Anjali','Karan','Meera','Vikram','Nisha','priya','Arjun','Swati','Naveen',
            'Priya','Karan','Deepa','Suresh','Sanjana'],
    'Age':[25,np.nan,30,25,28,22,24,np.nan,27,26,23,29,np.nan,25,22,28,24,np.nan,28,26,25,30,27,np.nan,25],
    'Marks':[85,90,np.nan,85,78,95,88,72,np.nan,91,67,84,79,np.nan,93,88,76,np.nan,99,85,71,72,89,99,75],
    'City':['Mysuru','Mysuru','Bengaluru','Mysuru',np.nan,'Delhi','Mysuru',
            'Manglore','Banglore',np.nan,'Mysuru','Hubli','Manglore',
            'Delhi','Bengaluru',np.nan,'Mysuru','Hubli',np.nan,'Delhi','Banglore',
            'Mysuru','Banglore','Manglore','Hubli']
}

df=pd.DataFrame(data)
print("Dataset:")
print(df)
```

```python
print("\nShape of dataset (rows,coloumns):",df.shape)
```

```python
print("\n Column name:",df.columns.tolist())
```

```python
print("\nData types of each coloumn:")
print(df.dtypes)
```

```python
print("\nDataset info:")
df.info()
```

```python
print("\nDescriptive statistics (numerical columns):")
print(df.describe())
```

```python
print("\nMean:\n",df.mean(numeric_only=True))
```

```python
print("\nMedian:\n",df.median(numeric_only=True))
```

```python
print("\nStandard deviation:\n",df.std(numeric_only=True))
```

```python
print("\nMissing value sin each column:")
print(df.isnull().sum())
print("\ntotal missing values in dataset:",df.isnull().sum().sum())
```
